```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

***Simulate dilation, erosion, opening and closing operation on images***







```python
image = cv2.imread('/content/fluid-black-white-patterns_1409-8819.avif', 0)
cv2_imshow(image)
```

```python
kernel = np.ones((5, 5), np.uint8)
```

```python
erosion = cv2.erode(image, kernel, iterations=1)
cv2_imshow(erosion)
```

```python
dilation = cv2.dilate(image, kernel, iterations=1)
cv2_imshow(dilation)
```

```python
opening = cv2.morphologyEx(image, cv2.MORPH_OPEN, kernel)
cv2_imshow(opening)
```

```python
closing = cv2.morphologyEx(image, cv2.MORPH_CLOSE, kernel)
cv2_imshow(closing)
```
